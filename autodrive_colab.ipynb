## 2) อัปโหลดสคริปต์ `autodrive.py`
เลือกไฟล์ `autodrive.py` จากเครื่องคุณ (อยู่ใน D:\AutoDrive)

## 3) (ทางเลือก) อัปโหลดวิดีโอของคุณเอง
ข้ามเซลล์นี้ได้ — ถ้าข้าม สคริปต์จะโหลดวิดีโอ dashcam ตัวอย่างให้เอง

```python
from google.colab import files
myvid = files.upload()   # เลือกไฟล์วิดีโอ .mp4 (ข้ามได้)
VIDEO = list(myvid.keys())[0] if myvid else None
print('วิดีโอ:', VIDEO or 'จะใช้ตัวอย่างอัตโนมัติ')
```

## 4) ประมวลผล (detect + ระยะ + overlay) แล้วเซฟเป็น out.mp4

```python
import subprocess
cmd = ['python', 'autodrive.py', '--save', 'out.mp4', '--no-show', '--max-width', '960']
try:
    if VIDEO:
        cmd += ['--source', VIDEO]
except NameError:
    pass
print('รัน:', ' '.join(cmd))
subprocess.run(cmd, check=True)
```

## 5) แสดงผลวิดีโอในหน้าเว็บ (แปลงเป็น H.264 ให้เบราว์เซอร์เล่นได้)

```python
import subprocess
from base64 import b64encode
from IPython.display import HTML
subprocess.run(['ffmpeg','-y','-i','out.mp4','-vcodec','libx264','-an','out_h264.mp4'], check=True)
data = b64encode(open('out_h264.mp4','rb').read()).decode()
HTML(f'<video width=720 controls><source src="data:video/mp4;base64,{data}" type="video/mp4"></video>')
```
